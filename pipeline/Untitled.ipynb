```python
import pandas as pd
from sqlalchemy import create_engine


engine = create_engine('postgresql://root:root@localhost:5432/ny_taxi')


df_green = pd.read_parquet('green_tripdata_2025-11.parquet')
df_zones = pd.read_csv('taxi_zone_lookup.csv')


print("importing ny taxi dataset...")
df_green.to_sql(name='green_taxi_data', con=engine, if_exists='replace')

print("importing regional dataset")
df_zones.to_sql(name='zones', con=engine, if_exists='replace')

print("importing all finished！")
```
